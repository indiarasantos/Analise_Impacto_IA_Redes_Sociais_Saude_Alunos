````python
import polars as pl
import time
````

````python
inicial = (
    pl.scan_parquet("estudantes.parquet"))

display(inicial.collect())
````

### Filtro alto uso de redes (>5h) e pouco sono (<=6h).
---

````python
# Filtro composto (E): muito tempo em redes sociais e pouco sono
q_risco = inicial.filter(
    (pl.col("horas_redes_sociais") > 5) & (pl.col("horas_sono") <= 6)
)

total = inicial.select(pl.len()).collect().item()
n_risco = q_risco.select(pl.len()).collect().item()

print(f"Filtro alto uso de redes (>5h) e pouco sono (<=6h): {n_risco:,} estudantes ({n_risco/total:.1%})")
display(q_risco.collect())
````

```
Análise: 10.5% dos estudantes da amostra, usam as redes por 5h ou mais e possuem 6h ou menos de sono.
```

### Filtro países desenvolvidos vs demais países.
---

````python
# Filtro com lista (IS IN): estudantes de países desenvolvidos x demais
q_desenvolvidos = inicial.filter(pl.col("nivel_desenvolvimento").is_in(["Desenvolvido"]))
q_demais = inicial.filter(pl.col("nivel_desenvolvimento").is_in(["Em desenvolvimento", "Subdesenvolvido"]))

print(f"Filtro países desenvolvidos: {q_desenvolvidos.select(pl.len()).collect().item():,}")
display(q_desenvolvidos.collect())
print(f"Filtro demais países: {q_demais.select(pl.len()).collect().item():,}")
display(q_demais.collect())
````

```
Análise: 149.548 estudantes da amostra, residem em países desenvolvidos, enquanto 350.452 dos estudantes residem em países subdesenvolvidos ou em desenvolvimento.
```

### Somente colunas redes sociais, bem-estar e capacidade de atenção.
---

````python
# Selecionar apenas as colunas do tema (redes sociais, bem-estar e capacidade de atenção)
cols_tema = ["id_estudante","horas_redes_sociais", "indice_bem_estar", "capacidade_atencao_minutos"]
q_tema = inicial.select(list(dict.fromkeys(cols_tema)))

# .drop() remove as colunas listadas
q_sem_infra = inicial.drop(["indice_infraestrutura_internet", "velocidade_media_internet_mbps"])

display(q_tema.head().collect())
display(q_sem_infra.head().collect())
print("q_tema:", q_tema.collect_schema().len(), "colunas | q_sem_infra:", q_sem_infra.collect_schema().len(), "colunas")
````

```
Análise: no recorte acima conseguimos visualizar com precisão os primeiros 5 estudantes.
```

### Categorização de horas de sono e uso de redes sociais.
---

````python
# Categorias de sono, uso de redes sociais
exprs = [
    pl.when(pl.col("horas_sono") <= 6).then(pl.lit("1-Baixo (≤6h)"))
      .when(pl.col("horas_sono") < 8).then(pl.lit("2-Médio (6-8h)"))
      .otherwise(pl.lit("3-Bom (≥8h)")).alias("categoria_sono"),

    pl.when(pl.col("horas_redes_sociais") < 2).then(pl.lit("1-Baixo (<2h)"))
      .when(pl.col("horas_redes_sociais") < 5).then(pl.lit("2-Moderado (2-5h)"))
      .otherwise(pl.lit("3-Alto (≥5h)")).alias("categoria_redes"),
]

q_cat = inicial.with_columns(exprs)
cols_cat = [c for c in ["horas_sono", "categoria_sono", "horas_redes_sociais", "categoria_redes"] if c]
display(q_cat.select(cols_cat).head(10).collect())
````

```
Análise: categorização de sono e uso de redes baseado na quantidade de horas.
```

### Pequenas amostras das primeiras 10 linhas e também da linhas 100 até a linha 150.
---

````python
q_head = inicial.head(10)          # primeiras 10 linhas
q_meio = inicial.slice(100, 50)    # offset 100, 50 linhas

display(q_head.collect())
display(q_meio.collect())
````

```
Análise: dividino a amostra em duas pequenas partes para usar como exemplo.
```

````python
# Recorte de linhas específicas (linhas 0, 10 e 50)
q_recorte = (
    inicial
    .with_row_index("id_linha")
    .filter(pl.col("id_linha").is_in([0, 10, 50]))
    .select(["id_estudante","pais", "nivel_desenvolvimento", "horas_redes_sociais", "horas_sono"])
)
display(q_recorte.collect())
````

```
Análise: recorte de um estudante aleatório de cada nível de desenvolvimento.
```

### Agrupamento por categoria de uso de redes e horas de sono.
---

````python
# Cruzamento: redes sociais x sono -> bem-estar e frequência
cruz = (
    q_cat.group_by(["categoria_redes", "categoria_sono"])
    .agg(
        pl.len().alias("estudantes"),
        pl.col("indice_bem_estar").mean().round(2).alias("media_bem_estar"),
    )
    .sort(["categoria_redes", "categoria_sono"])
)
display(cruz.collect())
````

```
Análise: conseguimos perceber a partir da amostra acima que, o alto uso de redes sociais impacta tanto na categoria do sono quanto na média de bem-estar.
```

````python
print("=== TESTE 1: Processamento SEM StringCache ===")
start_1 = time.time()

q_sem_cache = (
    pl.scan_parquet("estudantes.parquet")
    .group_by(["pais", "nivel_desenvolvimento"]) 
    .agg([
        pl.col("pontuacao_vicio_digital").mean().alias("media"),
        pl.col("pontuacao_vicio_digital").std().alias("desvio")
    ])
    .sort("media", descending=True)
)

df_1 = q_sem_cache.collect()
end_1 = time.time()

tempo_sem = end_1 - start_1

print(f"Tempo SEM Cache: {tempo_sem:.4f} segundos")
print(f"Uso de memória estimado (coluna pais): {df_1['pais'].estimated_size()} bytes\n")
````

````python
print("=== TESTE 2: Processamento COM StringCache ===")
# O StringCache funciona como um contexto (with)
with pl.StringCache():
    start_2 = time.time()

    q_com_cache = (
    pl.scan_parquet("estudantes.parquet")
    .group_by(["pais", "nivel_desenvolvimento"]) 
        .agg([
            pl.col("pontuacao_vicio_digital").mean().alias("media"),
            pl.col("pontuacao_vicio_digital").std().alias("desvio")
        ])
        .sort("media", descending=True)
    )

df_2 = q_com_cache.collect() # Ação!
end_2 = time.time()

tempo_com = end_2 - start_2

print(f"Tempo COM Cache: {tempo_com:.4f} segundos")
print(f"Uso de memória estimado (coluna pais): {df_2['pais'].estimated_size()} bytes\n")
````
