## Big Data e Processamento Otimizado com Python II

### Manipulação do Parquet e StringCache
---

````python
inicial = (
    pl.scan_parquet("estudantes.parquet"))

display(inicial.collect())
````

### Filtros
```
Para selecionar linhas baseadas em condições lógicas (semelhante ao WHERE do SQL).
```
---

````python
# Filtro Simples: Apenas estudantes de Singapura
q_singapura = inicial.filter(pl.col("pais") == "Singapura")

# Filtro Composto (E / OU): Singapura e idade maior que 20
q_singapura_pobreza = inicial.filter(
    (pl.col("pais") == "Singapura") & (pl.col("idade") > 20)
)

# Filtro com Lista (IS IN): Apenas paises desenvolvidos
q_desenvolvidos = inicial.filter(
    pl.col("nivel_desenvolvimento").is_in(["Desenvolvido"])
)

display(q_singapura.collect())
display(q_singapura_pobreza.collect())
display(q_desenvolvidos.collect())
````

### Seleção e Exclusão
```
Para escolher ou remover colunas específicas.
```
---

````python
# Selecionar apenas colunas de interesse
q_simples = inicial.select(["pais", "nivel_desenvolvimento", "idade"])

# .drop() remove as colunas listadas
q_sem_codigos = inicial.drop(["indice_infraestrutura_internet","velocidade_media_internet_mbps"])

# Seleciona tudo (implícito) excluindo as colunas listadas
q_sem_codigos = inicial.select(pl.exclude(["indice_infraestrutura_internet","velocidade_media_internet_mbps"]))

display(q_simples.collect())
display(q_sem_codigos.collect())
````

### Condicionais
```
Criar novas colunas com base em lógica (semelhante ao IF do Excel ou CASE WHEN do SQL).
```
---

````python
# Categorizar a saúde mental dos estudantes
q_categoria = inicial.with_columns(
    pl.when(pl.col("horas_sono") <= 6)
    .then(pl.lit("Baixo Sono"))
    .when(pl.col("horas_sono") < 8)
    .then(pl.lit("Médio Sono"))
    .otherwise(pl.lit("Bom Sono"))
    .alias("CATEGORIA_SONO")
)

display(q_categoria.select(["horas_sono", "CATEGORIA_SONO"]).collect())
````

````python
# Partindo do LazyFrame 'q_categoria' já criado
q_analise_categoria = (
    q_categoria
    .group_by("CATEGORIA_SONO")
    .agg([
        pl.len().alias("quantidade_estudantes"),
        pl.col("horas_sono").mean().alias("media_horas_sono"),
        pl.col("horas_sono").min().alias("menor_horas_sono"),
        pl.col("horas_sono").max().alias("maior_horas_sono"),
        pl.col("horas_sono").sum().alias("total_horas_sono")
    ])
    .sort("media_horas_sono", descending=True)
)

display(q_analise_categoria.collect())
````

### Fatiamento
```
Para pegar "pedaços" dos dados. No modo Lazy, o slice é muito eficiente pois o Polars lê apenas o necessário do disco.
```
---

````python
# Pegar as primeiras 10 linhas
q_head = inicial.head(10)

# Pegar da linha 100 até a 150 (offset 100, length 50)
q_meio = inicial.slice(100, 50)

display(q_head.collect())
display(q_meio.collect())
````

````python
# Recorte de linhas específicas (ex: linha 0, 10 e 50)
q_recorte_indices = (
    inicial
    .with_row_index("id_linha")  # Cria uma coluna temporária com o número da linha
    .filter(pl.col("id_linha").is_in([0, 10, 50])) # Filtra pelos números desejados
    .select(["pais", "nivel_desenvolvimento"]) # Seleciona as colunas
)

display(q_recorte_indices.collect())
````

### Agrupamento Avançado (group_by + agg)
---

````python
q_stats_mun = (
    inicial.group_by("nivel_escolaridade")
    .agg([
        pl.len().alias("total_estudantes"),  # Contagem rápida
        pl.col("horas_redes_sociais").sum().alias("horas_redes_sociais_total"),  # Soma
        pl.col("taxa_frequencia_aulas").mean().alias("media_frequencia"), # Média
        pl.col("capacidade_atencao_minutos").mean().alias("media_atencao"),  # Média
        
    ])
    .sort("horas_redes_sociais_total", descending=True)
)

display(q_stats_mun.collect())
````

### StringCache e Otimização de CPU
---

````python
print("=== TESTE 1: Processamento SEM StringCache ===")
start_1 = time.time()

q_sem_cache = (
    pl.scan_parquet("estudantes.parquet")
    .group_by(["pais", "nivel_desenvolvimento"]) 
    .agg([
        pl.col("pontuacao_vicio_digital").mean().alias("media"),
        pl.col("pontuacao_vicio_digital").std().alias("desvio")
    ])
    .sort("media", descending=True)
)

df_1 = q_sem_cache.collect()
end_1 = time.time()

tempo_sem = end_1 - start_1

print(f"Tempo SEM Cache: {tempo_sem:.4f} segundos")
print(f"Uso de memória estimado (coluna pais): {df_1['pais'].estimated_size()} bytes\n")
````

````python
print("=== TESTE 2: Processamento COM StringCache ===")
# O StringCache funciona como um contexto (with)
with pl.StringCache():
    start_2 = time.time()

    q_com_cache = (
    pl.scan_parquet("estudantes.parquet")
    .group_by(["pais", "nivel_desenvolvimento"]) 
        .agg([
            pl.col("pontuacao_vicio_digital").mean().alias("media"),
            pl.col("pontuacao_vicio_digital").std().alias("desvio")
        ])
        .sort("media", descending=True)
    )

df_2 = q_com_cache.collect() # Ação!
end_2 = time.time()

tempo_com = end_2 - start_2

print(f"Tempo COM Cache: {tempo_com:.4f} segundos")
print(f"Uso de memória estimado (coluna pais): {df_2['pais'].estimated_size()} bytes\n")
````
