## 🎯 Pergunta e escopo

**Pergunta:** o uso de redes sociais está associado à **saúde** (sono e bem-estar) e ao **desempenho** (frequência às aulas, capacidade de atenção e risco acadêmico) dos estudantes?

**Delimitação do tema:** uso de redes sociais (horas por dia e vício digital), saúde, desempenho. Uso de IA e notas **não fazem parte do escopo**.

---
## 💡 Resposta em uma frase

> Quanto mais tempo em redes sociais, **menos sono e menor bem-estar** (associação forte), e **menor atenção e frequência** (associação mais fraca). O sono aparece como o fator de maior peso para o bem-estar.

---
## 🧩 O que cada etapa mostrou

| Etapa | Pergunta | Principal resultado |
|---|---|---|
| **Parte 1** | Como se comportam o bem-estar e o sono por faixa de uso? | Bem-estar médio de **74,11** (pouco uso, sono bom) contra **45,15** (alto uso, pouco sono); **10,5%** dos estudantes estão nesse grupo de risco |
| **Parte 2** | Como é a distribuição do uso de redes? | Média de **3,36 h/dia**, distribuição simétrica; faixa *Alto* (≥ 5 h) com cerca de **11%** e *outliers* (> 6,81 h) com **0,7%** |
| **Correlações** | Qual a força das relações? | **Forte** com sono (-0,76) e bem-estar (-0,63), **moderada** com atenção (-0,43), **fraca** com frequência (-0,28) e risco acadêmico (+0,18) |
| **Regressão** | Quanto cada hora em redes explica e pesa? | Redes explicam **58%** do sono e **40%** do bem-estar; com o sono no modelo, ele passa a pesar mais que as redes no bem-estar |

---
## 📊 Resultados integrados

| Dimensão | Correlação com horas em redes | R² da regressão | Efeito de +1 h em redes |
|---|---|---|---|
| 😴 Sono | **-0,76** (forte) | 0,58 | cerca de **34 min a menos** |
| 💚 Bem-estar | **-0,63** (forte) | 0,40 (0,51 com o sono) | **-5,2 pontos** (-2,0 com o sono constante) |
| 🎯 Atenção | **-0,43** (moderada) | 0,20 | cerca de **1,3 min a menos** |
| 🏫 Frequência às aulas | **-0,28** (fraca) | 0,08 | cerca de **1 ponto percentual a menos** |
| 📉 Risco acadêmico | **+0,18** (fraca) | não modelado | — |

*Para ilustrar: passar de 2 h para 5 h por dia em redes se associa, em média, a cerca de 1,7 h a menos de sono e 15 pontos a menos de bem-estar (modelos simples). É uma associação, não uma causa.*

---
## 🔗 Fio condutor

1. **Gradiente da saúde para o desempenho:** a associação é forte onde o impacto é mais direto (sono e bem-estar) e vai diminuindo até o desempenho (atenção, frequência e risco acadêmico). O desempenho depende de muitos outros fatores.
2. **O sono é central:** redes e sono andam juntos (r = -0,76; 97,8% dos usuários intensos dormem 6 h ou menos). Ao incluir o sono, o efeito das redes sobre o bem-estar cai cerca de 62%. Isso é **compatível** com o sono ser parte do caminho entre redes e bem-estar, mas não prova isso.
3. **Quem está em risco:** cerca de **1 em cada 10** estudantes combina uso intenso e pouco sono, e o uso estatisticamente atípico atinge **0,7%**.
