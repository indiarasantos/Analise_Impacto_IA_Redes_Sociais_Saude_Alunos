```python
import polars as pl

inicial = (
    pl.scan_parquet("estudantes.parquet"))
    
display(inicial.collect())
```

```python
import polars as pl

telas = ["horas_redes_sociais", "horas_videos_curtos", "sessoes_por_dia",
         "duracao_media_sessao_minutos", "pontuacao_madrugada"]

efeitos = ["horas_sono", "nivel_estresse", "pontuacao_ansiedade",
           "pontuacao_produtividade", "indice_bem_estar", "indice_brain_rot"]

df = inicial.select(telas + efeitos).drop_nulls().collect()

resultado = df.select([
    pl.corr(t, e).alias(f"{t} x {e}")
    for t in telas for e in efeitos
])

# vira tabela: linhas = telas, colunas = efeitos
tabela = (
    resultado.transpose(include_header=True, column_names=["r"])
    .with_columns(pl.col("r").round(2))
)
display(tabela)
```

```python
import polars as pl
import seaborn as sns
import matplotlib.pyplot as plt

df = inicial.select(telas + efeitos).drop_nulls().collect()

# matriz de correlação em Polars (linhas na mesma ordem das colunas)
corr = df.corr().with_columns(pl.Series("variavel", df.columns))

# só telas (linhas) x efeitos (colunas)
corr_xy = (
    corr
    .filter(pl.col("variavel").is_in(telas))
    .select(["variavel"] + efeitos)
)

sns.heatmap(
    corr_xy.select(efeitos).to_numpy(),
    annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1,
    xticklabels=efeitos,
    yticklabels=corr_xy["variavel"].to_list(),
)
plt.title("Uso de telas × saúde e desempenho")
plt.tight_layout()
plt.show()
```
